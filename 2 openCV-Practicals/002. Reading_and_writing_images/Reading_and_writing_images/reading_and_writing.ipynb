```python
import cv2
```

```python
# Read the image, this store as numpy array
image = cv2.imread("./mountain.jpg")
print(image)
```

```python
# check if image is loaded successfully

if image is None:
    print("Error : Cound not load the image.")
else:
    print("Image loaded successfully.")
```

```python
from IPython.display import Image, display
display(image)

if image is None:
    print("Error: Could not load the image.")
else:
    _, buffer = cv2.imencode(".jpg", image)
    display(Image(data=buffer.tobytes()))

# cv2.imshow("Image window", image)
# cv2.waitKey(6000) -> closes automatically after 6000 milliseconds (6 seconds)
# cv2.destroyAllWindows() -> cloe all opencv window
# cv2.waitKey(0) -> if any keypressed then only closed

# Note : This cv.show() crash mac , by opening gui exernally to show img that's why i used ipython lib
```

```python
# Write the image

image_new = image + 30 # add 30 to all pixel
cv2.imwrite("./output.jpg", image_new) # save image as output.jpg and for image_new is there in output
```

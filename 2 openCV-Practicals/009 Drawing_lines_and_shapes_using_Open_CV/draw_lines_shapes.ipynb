```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Create a empty canvas
height, width = 500, 700
blue = (255, 127, 0) # B, G , R
canvas = np.full((height, width, 3), blue, dtype=np.uint8)

plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB)) # as matplotlib want image to be in RGB 
```

# Draw to a line

```python
cv2.line(canvas, (50, 50), (650, 450), (255, 255, 255), 5) # p1 corr, p2 corr, color BGR - white , thickness

plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```

# Draw a rectangle

```python
cv2.rectangle(canvas, (100, 100), (300, 300), (0, 255, 255), -1) # -1 means fill whole rectangle

plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```

# Draw a circle

```python
cv2.circle(canvas, (500, 150), 100, (0, 0, 255), 5) # 100 is radius

plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```

# Draw an Ellipse

```python
cv2.ellipse(canvas, (350, 350), (100, 50), 45, 0, 360, (0, 255, 0), 3)

plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```

# Draw a polygon

```python
# Xn, Yn, where n>2 => called polygons

points = np.array([[150, 400], [300, 400], [225, 300]], np.int32)
new_reshaped_point = points.reshape((-1, 1, 2))

print(new_reshaped_point)


cv2.polylines(canvas, [new_reshaped_point], False, (255, 0, 255), 3) # if true then ploygons will closed 

plt.imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB))
```
